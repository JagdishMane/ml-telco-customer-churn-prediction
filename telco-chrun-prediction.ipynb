# Project 

### Business Context
---

Telco companies face ~15~23% annual churn; retaining a customere costs significantly less than acquiring a new one. The goal is to predict which customeres are likely to churn so taht proactive retention campaigns can be targeted.

### Dataset
---

**Synthetic Telco Churn 100K Rows**

This is a fully synthetic, high-quality dataset designed for customer churn prediction in the telecommunications (telco) domain — perfect for machine learning practice, portfolio projects, Kaggle notebooks, and interviews.

With 100,000 realistic rows, it mimics real-world telco customer behavior while avoiding any privacy concerns. Features are carefully engineered to reflect typical telecom billing, contract, and demographic patterns, including realistic imbalance in churn (~20% positive class for challenging classification tasks).
